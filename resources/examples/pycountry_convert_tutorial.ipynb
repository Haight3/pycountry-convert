# Getting started with `pycountry-convert-ng`

This short tutorial shows how to convert between country names, ISO 3166-1 alpha-2 and alpha-3 codes, and continents. The package is installed as `pycountry-convert-ng` but imported as `pycountry_convert`.

## Installation

Install the maintained distribution before running the notebook:

```bash
pip install pycountry-convert-ng
```

````python
import pycountry_convert as pc

print(f"pycountry-convert-ng version: {pc.__version__}")
````

## Convert country names and ISO codes

The package exposes direct functions for the common country-name and ISO-code conversions. ISO codes should be uppercase.

````python
country_name = "Germany"
alpha2 = pc.country_name_to_country_alpha2(country_name)
alpha3 = pc.country_name_to_country_alpha3(country_name)

print(f"{country_name}: alpha-2={alpha2}, alpha-3={alpha3}")
print(f"{alpha2} resolves to {pc.country_alpha2_to_country_name(alpha2)}")
print(f"{alpha3} resolves to alpha-2 code {pc.country_alpha3_to_country_alpha2(alpha3)}")
````

Common alternative names are also supported, including aliases such as `South Korea` and `Great Britain`.

````python
for country_name in ["South Korea", "Great Britain", "Ivory Coast"]:
    alpha2 = pc.country_name_to_country_alpha2(country_name)
    print(f"{country_name}: {alpha2}")
````

## Find a country's continent

Continent lookup uses two steps: convert the country to an alpha-2 code, then convert that code to a continent code. A final helper converts the continent code to its display name.

````python
country_name = "Japan"
alpha2 = pc.country_name_to_country_alpha2(country_name)
continent_code = pc.country_alpha2_to_continent_code(alpha2)
continent_name = pc.convert_continent_code_to_continent_name(continent_code)

print(f"{country_name} ({alpha2}) is in {continent_name} ({continent_code})")
````

## Control country-name formatting

When converting an alpha-2 code to a country name, pass one of the package's formatting constants as the second argument.

````python
print(pc.country_alpha2_to_country_name("DE"))
print(pc.country_alpha2_to_country_name("DE", pc.COUNTRY_NAME_FORMAT_LOWER))
print(pc.country_alpha2_to_country_name("DE", pc.COUNTRY_NAME_FORMAT_UPPER))
````

## Convert several countries

The functions are small building blocks, so they can be composed in an ordinary loop without additional dependencies.

````python
countries = ["Brazil", "Germany", "Japan", "South Africa"]

for country_name in countries:
    alpha2 = pc.country_name_to_country_alpha2(country_name)
    continent_code = pc.country_alpha2_to_continent_code(alpha2)
    continent_name = pc.convert_continent_code_to_continent_name(continent_code)
    print(f"{country_name:12} | {alpha2} | {continent_name}")
````

## Handle invalid input

Unknown country names and codes raise `KeyError`. Catch it when input comes from users or another untrusted source.

````python
try:
    pc.country_alpha2_to_country_name("ZZ")
except KeyError as error:
    print(error)
````
